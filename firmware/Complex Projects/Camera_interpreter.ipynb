```python
import serial
import time
import numpy as np

arduino = serial.Serial(port='COM10', baudrate=115200, timeout=1)

def arduino_read():
    #arduino.write(bytes(x + '\n', 'utf-8'))
    time.sleep(1)
    print("start")
    data = arduino.readline()
    print("end")
    try:
        return data.decode('utf-8')[:-1]
    except:
        return []
```

```python
while (True):
    data = arduino_read()
    
    print(data[:8])
    
    if data[0:8] == "Ping :D ":
        with open("data.txt", "w") as my_file:
            my_file.write(data[8:])
        print("Finished!")
        break
```

```python
import numpy as np
from PIL import Image

with open("data.txt", "r") as my_file:
    data = my_file.readline()
    data = data.strip().split()
    data = np.uint32(np.asarray(data))
    data = [[(_ >> 16) & 0xFF, (_ >> 8) & 0xFF, _ & 0xFF] for _ in data]
    #data = np.uint8(np.asarray(data).reshape((720, 1280, 3)))
    data = np.uint8(np.asarray(data).reshape((72, 128, 3)))
    
    #img = Image.fromarray(data , 'L')
    #img.show()
    img = Image.fromarray(data, 'RGB')
    img.show()
```
